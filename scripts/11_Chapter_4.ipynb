- Forecast evaluation is conducted using an expanding-window walk-forward procedure beginning with a 60-month initial estimation period. At each step, the model is re-estimated using all available historical data up to time t−1 and used to generate a one-step-ahead prediction for the forward 24-month correlation at time t.

- The forward 24-month correlation is constructed by first computing a backward 24-month rolling correlation and then shifting the series backward by 23 months so that the value at time t reflects the realized correlation over months t through t+23.

- The forward-looking target naturally truncates the final 23 months of the sample, as full future return realizations are required to compute the 24-month correlation window.

```python
def run_rolling_forecast(df, features, target, model_name="Model"):
    initial_train_size = 60
    actuals, predictions, dates = [], [], []

    for i in range(initial_train_size, len(df) - 1):
        train = df.iloc[:i]
        test  = df.iloc[i:i+1]

        model = GradientBoostingRegressor(
            n_estimators=400,
            learning_rate=0.01,
            max_depth=5,
            subsample=0.8,
            random_state=42
        )
        model.fit(train[features], train[target])
        pred = model.predict(test[features])

        actuals.append(test[target].values[0])
        predictions.append(pred[0])
        dates.append(test.index[0])

    # --- Convert to arrays ---
    y_true = np.asarray(actuals, dtype=float)
    y_pred = np.asarray(predictions, dtype=float)

    # --- Metrics ---
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    mae  = mean_absolute_error(y_true, y_pred)

    r_val, r_p = pearsonr(y_true, y_pred)

    def _fisher_z(x): 
        return np.arctanh(np.clip(x, -0.999999, 0.999999))
    z_rmse = np.sqrt(mean_squared_error(_fisher_z(y_true), _fisher_z(y_pred)))

    def _direction(series):
        return np.array([1 if j - i > 0 else 0 for i, j in zip(series[:-1], series[1:])])
    actual_dir    = _direction(y_true)
    predicted_dir = _direction(y_pred)

    trend_acc = accuracy_score(actual_dir, predicted_dir)
    f1_dir    = f1_score(actual_dir, predicted_dir)

    # --- Confusion Matrix & Classification Report ---
    cm = confusion_matrix(actual_dir, predicted_dir)
    report_df = pd.DataFrame(
        classification_report(actual_dir, predicted_dir,
                              target_names=["Down", "Up"], output_dict=True)
    ).transpose().round(2)

    # --- Return results for ablation summary ---
    return {
        "metrics": {
            "rmse": rmse,
            "mae": mae,
            "fisher_z_rmse": z_rmse,
            "pearson_r": r_val,
            "pearson_p": r_p,
            "trend_accuracy": trend_acc,
            "f1_direction": f1_dir
        },
        "confusion_matrix": cm,
        "report_df": report_df,
        "y_true": y_true,
        "y_pred": y_pred,
        "dates": dates
    }
```

```python
def run_rolling_forecast_series(df, features, target, model_name="Model"):
    initial_train_size = 60
    actuals, predictions, dates = [], [], []

    for i in range(initial_train_size, len(df) - 1):
        train = df.iloc[:i]
        test  = df.iloc[i:i+1]

        model = GradientBoostingRegressor(
            n_estimators=400,
            learning_rate=0.01,
            max_depth=5,
            subsample=0.8,
            random_state=42
        )
        model.fit(train[features], train[target])
        pred = model.predict(test[features])

        actuals.append(test[target].values[0])
        predictions.append(pred[0])
        dates.append(test.index[0])

    # --- Convert to arrays ---
    y_true = np.asarray(actuals, dtype=float)
    y_pred = np.asarray(predictions, dtype=float)

    # --- Metrics ---
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    mae  = mean_absolute_error(y_true, y_pred)

    r_val, r_p = pearsonr(y_true, y_pred)

    def _fisher_z(x): 
        return np.arctanh(np.clip(x, -0.999999, 0.999999))
    z_rmse = np.sqrt(mean_squared_error(_fisher_z(y_true), _fisher_z(y_pred)))

    def _direction(series):
        return np.array([1 if j - i > 0 else 0 for i, j in zip(series[:-1], series[1:])])
    actual_dir    = _direction(y_true)
    predicted_dir = _direction(y_pred)

    trend_acc = accuracy_score(actual_dir, predicted_dir)
    f1_dir    = f1_score(actual_dir, predicted_dir)

    # --- Confusion Matrix & Classification Report ---
    cm = confusion_matrix(actual_dir, predicted_dir)
    report_df = pd.DataFrame(
        classification_report(actual_dir, predicted_dir,
                              target_names=["Down", "Up"], output_dict=True)
    ).transpose().round(2)

    # --- Return results for ablation summary ---
    return {
        "metrics": {
            "rmse": rmse,
            "mae": mae,
            "fisher_z_rmse": z_rmse,
            "pearson_r": r_val,
            "pearson_p": r_p,
            "trend_accuracy": trend_acc,
            "f1_direction": f1_dir
        },
        "predictions": pd.Series(y_pred, index=dates),
        "actuals": pd.Series(y_true, index=dates)
    }
```

```python
import numpy as np
from scipy import stats
import pandas as pd

def diebold_mariano_test_from_results(results1, results2, h=24):
    actual1 = pd.Series(results1["y_true"], index=results1["dates"])
    pred1   = pd.Series(results1["y_pred"], index=results1["dates"])

    actual2 = pd.Series(results2["y_true"], index=results2["dates"])
    pred2   = pd.Series(results2["y_pred"], index=results2["dates"])

    common_index = actual1.index.intersection(actual2.index)
    common_index = common_index.intersection(pred1.index).intersection(pred2.index)

    y = actual1.loc[common_index].values
    p1 = pred1.loc[common_index].values
    p2 = pred2.loc[common_index].values

    # squared-error loss
    e1 = (y - p1) ** 2
    e2 = (y - p2) ** 2
    d = e1 - e2

    T = len(d)
    d_bar = np.mean(d)

    # HAC variance estimate
    gamma0 = np.var(d, ddof=1)
    hac_var = gamma0

    for lag in range(1, h):
        cov = np.cov(d[lag:], d[:-lag], ddof=1)[0, 1]
        weight = 1 - lag / h
        hac_var += 2 * weight * cov

    dm_stat = d_bar / np.sqrt(hac_var / T)
    p_value = 2 * (1 - stats.norm.cdf(abs(dm_stat)))

    return {
        "dm_stat": dm_stat,
        "p_value": p_value,
        "mean_loss_diff": d_bar,
        "n_obs": T
    }

dm_result = diebold_mariano_test_from_results(
    baseline_results,
    sentiment_results,
    h=24
)

print("DM statistic:", round(dm_result["dm_stat"], 4))
print("p-value:", round(dm_result["p_value"], 4))
print("Mean loss differential:", round(dm_result["mean_loss_diff"], 6))
print("N observations:", dm_result["n_obs"])
```

```python
def run_rolling_forecast_regime_split(
    df,
    features,
    target,
    regime_col="Regime",
    model_name="Regime-Split Model",
    initial_train_size=60,
    min_regime_train_size=20
):
    actuals, predictions, dates, regimes_used = [], [], [], []

    df = df.dropna().copy()

    for i in range(initial_train_size, len(df) - 1):
        train = df.iloc[:i]
        test  = df.iloc[i:i+1]

        test_regime = test[regime_col].iloc[0]

        # --- Chen-style: train ONLY on same regime ---
        train_regime = train[train[regime_col] == test_regime]

        # Skip if not enough data in this regime
        if len(train_regime) < min_regime_train_size:
            continue

        model = GradientBoostingRegressor(
            n_estimators=400,
            learning_rate=0.01,
            max_depth=5,
            subsample=0.8,
            random_state=42
        )

        model.fit(train_regime[features], train_regime[target])
        pred = model.predict(test[features])

        actuals.append(test[target].values[0])
        predictions.append(pred[0])
        dates.append(test.index[0])
        regimes_used.append(test_regime)

    # --- Convert ---
    y_true = np.asarray(actuals, dtype=float)
    y_pred = np.asarray(predictions, dtype=float)

    # --- Metrics ---
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    mae  = mean_absolute_error(y_true, y_pred)

    r_val, r_p = pearsonr(y_true, y_pred)

    def _fisher_z(x):
        return np.arctanh(np.clip(x, -0.999999, 0.999999))

    z_rmse = np.sqrt(mean_squared_error(_fisher_z(y_true), _fisher_z(y_pred)))

    def _direction(series):
        return np.array([1 if j - i > 0 else 0 for i, j in zip(series[:-1], series[1:])])

    actual_dir = _direction(y_true)
    predicted_dir = _direction(y_pred)

    trend_acc = accuracy_score(actual_dir, predicted_dir)
    f1_dir    = f1_score(actual_dir, predicted_dir)

    return {
        "metrics": {
            "rmse": rmse,
            "mae": mae,
            "fisher_z_rmse": z_rmse,
            "pearson_r": r_val,
            "pearson_p": r_p,
            "trend_accuracy": trend_acc,
            "f1_direction": f1_dir,
            "n_obs": len(y_true)
        },
        "predictions": pd.Series(y_pred, index=dates),
        "actuals": pd.Series(y_true, index=dates),
        "regimes": pd.Series(regimes_used, index=dates)
    }
```

```python
# Same dates used by Chen-style regime split
chen_idx = chen_results["actuals"].index

def get_actual_pred_series(result):
    if "actuals" in result and "predictions" in result:
        y_true = result["actuals"]
        y_pred = result["predictions"]
    else:
        y_true = pd.Series(result["y_true"], index=result["dates"])
        y_pred = pd.Series(result["y_pred"], index=result["dates"])
    return y_true, y_pred


def subset_metrics(result, idx):
    y_true_all, y_pred_all = get_actual_pred_series(result)

    common_idx = y_true_all.index.intersection(y_pred_all.index).intersection(idx)

    y_true = y_true_all.loc[common_idx]
    y_pred = y_pred_all.loc[common_idx]

    return {
        "rmse": np.sqrt(mean_squared_error(y_true, y_pred)),
        "mae": mean_absolute_error(y_true, y_pred),
        "pearson_r": pearsonr(y_true, y_pred)[0],
        "n_obs": len(common_idx)
    }


same_sample_compare = pd.DataFrame({
    "Macro-Only": subset_metrics(baseline_results, chen_idx),
    "Macro + Sentiment": subset_metrics(sentiment_results, chen_idx),
    "Macro + Sentiment + Regime": subset_metrics(regime_results, chen_idx),
    "Chen-Style Regime Split": {
        "rmse": chen_results["metrics"]["rmse"],
        "mae": chen_results["metrics"]["mae"],
        "pearson_r": chen_results["metrics"]["pearson_r"],
        "n_obs": chen_results["metrics"]["n_obs"]
    }
}).T

same_sample_compare.round(4)
```

To further examine regime-based modeling approaches, a regime-splitting specification following Chen et al. (2024) is implemented, where separate models are estimated within each regime and used for prediction. The results reveal a trade-off between predictive specialization and data efficiency. The regime-split model demonstrates strong performance within the restricted sample where sufficient regime-specific observations are available, and it achieves improved directional accuracy, indicating enhanced sensitivity to short-term movements in correlation dynamics.

However, this approach requires discarding a substantial portion of the sample due to insufficient observations within certain regimes, which reduces its applicability in real-time forecasting contexts. The resulting reduction in effective sample size limits the model’s ability to learn stable relationships across time and introduces practical constraints on implementation.

In contrast, the proposed unified framework retains the full sample and incorporates regime information as an explanatory variable, allowing the model to capture regime-dependent effects without sacrificing data efficiency. This approach provides more stable and scalable performance while preserving the ability to model nonlinear dynamics. These findings suggest that while regime-specific modeling can enhance local predictive performance, integrating regime information within a unified framework offers a more balanced and practically robust approach to forecasting stock–bond correlation. This distinction highlights the importance of balancing model specialization with data availability in empirical forecasting settings.
